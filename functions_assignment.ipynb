```python
#Given a list of email strings:
#Requirements
#Use filter().
#A basic valid email should:
#Contain exactly one @.
#Have characters before @.
#Contain a dot after @.
#Return only emails satisfying these conditions
emails=["abhi@gmail.com","kartik@gmail.com","john@gmailcom"]
def is_valid_email(emails):
  if emails.count("@") !=1:
    return False
  name,domain=emails.split("@")
  if len(name)==0:
    return False
  if "." not in domain:
    return False
  return True
valid_emails=filter(is_valid_email,emails)
print(list(valid_emails))



```

```python
#Given dates in DD-MM-YYYY format:

#Requirements
#Ask the user for a year.
#Use filter() to return dates belonging to that year.
#Handle invalid date strings safely.
#Use datetime.strptime() for validation.
import datetime
dates=["15-03-2023","12-01-2024","25-12-2023","32-12-2022","10-05-2024"]
def belongs_to_year(date_str,target_year):
  try:
    date_obj=datetime.datetime.strptime(date_str.strip(),"%d-%m-%Y")
    return date_obj.year==target_year
  except ValueError:
    return False
year=int(input("enter a year"))
filtered_dates=list(filter(lambda d:belongs_to_year(d,year),dates))
print("Dates in",year,":",filtered_dates)
```

```python
#Write a function that calculates factorial using reduce().

#Requirements
#Accept a non-negative integer.
#Generate numbers from 1 to the given number.
#Multiply them using reduce().
#Return 1 when the input is 0.
#Reject negative values.
from functools import reduce
def factorial(n):
  if n < 0:
    return "invalid number"

  return reduce(lambda x, y: x * y, range(1, n + 1), 1)

number_input = input("enter a number")
try:
  number = int(number_input) # Convert input string to int here
  print(number, factorial(number))
except ValueError:
  print("Error: invalid input")
```

```python
def create_bank_account():
  balance = 0

  def deposit(amount):
    nonlocal balance
    balance += amount
    print(f"Deposited: {amount}. New balance: {balance}")

  def withdraw(amount):
    nonlocal balance
    if amount > balance:
      print("Insufficient balance: cannot withdraw {amount}. Current balance: {balance}")
    else:
      balance -= amount
      print(f"Withdrew: {amount}. New balance: {balance}")

  def balance_enquiry():
    return balance

  return deposit, withdraw, balance_enquiry

# Example usage:
deposit_func, withdraw_func, enquiry_func = create_bank_account()

deposit_func(1000)
withdraw_func(400)
print(f"Current balance: {enquiry_func()}")
withdraw_func(700) # Test insufficient balance
print(f"Final balance: {enquiry_func()}")
```

```python
#Create a decorator named execution_logger().

#Requirements
#The decorator should print a message before the original function executes.
#It should print another message after execution.
#Apply it to an addition function.
#Preserve the original function's return value.
def execution_logger(func):
  def wrapper(*args,**kwargs):
    print("starting function")
    result=func(*args,**kwargs)
    print("function completed")
    return result
  return wrapper
@execution_logger
def add(a,b):
  return a+b
print("result",add(10,20))
```

```python
#Develop a complete Employee Performance Analysis System using functions, lambda expressions, map(), filter(), reduce(), nested functions, closures, and LEGB scope.
employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Anitha",
        "department": "HR",
        "salary": 45000,
        "performance_score": 72,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Kiran",
        "department": "IT",
        "salary": 75000,
        "performance_score": 94,
        "experience": 7
    },
    {
        "id": 104,
        "name": "Meena",
        "department": "Finance",
        "salary": 55000,
        "performance_score": 81,
        "experience": 6
    },
    {
        "id": 105,
        "name": "Arjun",
        "department": "IT",
        "salary": 40000,
        "performance_score": 65,
        "experience": 2
    }
]

#Write a function named display_employees() that prints all employee details in a readable format.
def display_employees(list):
  for emp in list:
    print(f"id:{emp["id"]} | name:{emp["name"]}| department:{emp["department"]} | salary:{emp["salary"]}| performance_score:{emp["performance_score"]}| experience:{emp["experience"]}")
print(display_employees(employees))

#: Extract Employee Names
employee_names=list(map(lambda emp:emp["name"],employees))
print(employee_names)



#Use filter() to select employees whose performance score is 80 or above.
high_performance=list(filter(lambda emp:emp["performance_score"]>=80,employees))
print(high_performance)

#Write a function that accepts a department name and returns matching employees using filter().
department_name=list(filter(lambda emp:emp["department"]=="IT",employees))
print(department_name)

#Use map() to calculate revised salaries based on these conditions:

#Performance score 90 or above: 15% increase
#Performance score 80–89: 10% increase
#Performance score 70–79: 5% increase
#Below 70: No increase
def calculate_revised_salary(employee):
  if employee["performance_score"]>=90:
    return employee["salary"]*1.15
  elif employee["performance_score"]>=80:
    return employee["salary"]*1.10
  elif employee["performance_score"]>=70:
    return employee["salary"]*1.05
  else:
    return employee["salary"]*0.0
revised_salary=list(map(calculate_revised_salary,employees))
print(revised_salary)




```

```python
#Create an outer function named generate_department_report().

#Inside it, define nested functions to calculate:

#Number of employees
#Total salary
#Average salary
#Average performance score
#Return the complete department report as a dictionary.
employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Anitha",
        "department": "HR",
        "salary": 45000,
        "performance_score": 72,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Kiran",
        "department": "IT",
        "salary": 75000,
        "performance_score": 94,
        "experience": 7
    },
    {
        "id": 104,
        "name": "Meena",
        "department": "Finance",
        "salary": 55000,
        "performance_score": 81,
        "experience": 6
    },
    {
        "id": 105,
        "name": "Arjun",
        "department": "IT",
        "salary": 40000,
        "performance_score": 65,
        "experience": 2
    }
]
def generate_department_report(department_name,department_employees):
  def get_employee_count():
    return len(department_employees)
  def calculate_total_salary():
    return sum(emp["salary"] for emp in department_employees)
  def get_avg_salary():
    count = get_employee_count()
    if count == 0:
      return 0
      return round (calculate_total_salary()/count,2)

  def get_avg_performance():
    count= get_employee_count()
    if count==0:
      return 0
      total_score=sum(emp["performance_score"]for emp in department_employees)
      return round(total_score/count,2)
  report={
        "department":department_name,
        "number of employees":get_employee_count(),
        "total salary":calculate_total_salary(),
        "average salary":get_avg_salary(),
        "average performance score":get_avg_performance()
  }
  return report
departments={}
for emp in employees:
  dept=emp["department"]
  if dept not in departments:
    departments[dept]=[]
  departments[dept].append(emp)
print("department reports")
for department_name,department_emp_list in departments.items():
  report=generate_department_report(department_name,department_emp_list)
  print(report)


```
