```python
#Using filter(), map(), and reduce():

#Extract temperatures greater than 30°C.
#Convert the selected values into Fahrenheit.
#Calculate the average Fahrenheit temperature.
from functools import reduce
temperatures = [18, 24, 30, 35, 40, 12, 28, 33, 15, 42]
temperatures_30= lambda x:x>30
temperature_30_2=filter(temperatures_30,temperatures)
temperature_fah=map(lambda x:x*9/5+32,temperature_30_2)
temperature_average=reduce(lambda x,y:(x+y)/2,temperature_fah)
print(temperature_average)
```

```python
#Using filter(), map(), and reduce():

#Extract prices greater than 1000.
#Apply a 10% discount to every selected price.
#Calculate the total discounted amount.
from functools import reduce
prices = [450, 1200, 2500, 800, 5000, 1500, 7500]
price_high= lambda x:x>1000
price_high_2=filter(price_high,prices)
price_discount=map(lambda x:x*0.10,price_high_2)
total_discount=reduce(lambda x,y:x+y,price_discount)
print(total_discount)
```

```python
#Using filter() and reduce():

#Extract students who passed.
#Find the topper among the passed students.
students = [
    {"name": "Ravi", "marks": 92, "passed": True},
    {"name": "Kiran", "marks": 68, "passed": True},
    {"name": "Meena", "marks": 85, "passed": True},
    {"name": "Ajay", "marks": 48, "passed": True},
    {"name": "Sneha", "marks": 35, "passed": False},
    {"name": "Rahul", "marks": 95, "passed": True}
]
from functools import reduce
students_passed=lambda x:x["passed"]==True
students_passed2=filter(students_passed,students)
topper=reduce(lambda x,y:x if x["marks"]>y["marks"] else y,students_passed2)
print(topper)
```

```python
#Using filter(), map(), and reduce():

#Extract students scoring more than 70.
#Extract only their marks.
#Calculate the average marks.
students = [
    {"name": "Ravi", "marks": 92},
    {"name": "Kiran", "marks": 68},
    {"name": "Meena", "marks": 85},
    {"name": "Ajay", "marks": 48},
    {"name": "Sneha", "marks": 79},
    {"name": "Rahul", "marks": 95}
]
from functools import reduce
students_marks= lambda x:x["marks"]>70
students_marks_2=filter(students_marks,students)
students_only_marks=map(lambda x:x["marks"],students_marks_2)
students_average_marks=reduce(lambda x,y:x+y,students_only_marks)
students_average_marks2=students_average_marks/len(students)
print(students_average_marks2)
```

```python
#Using filter(), map(), and reduce():

#Extract products currently in stock.
#Calculate each product's inventory value using price × stock.
#Calculate the total inventory value.
products = [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]
from functools import reduce
products_stocks= lambda x:x["stock"]>0
products_in_stock=filter(products_stocks,products)
products_inventory=map(lambda x:x["price"]*x['stock'],products_in_stock)
product_total=reduce(lambda x,y:x+y,products_inventory)
print(product_total)
```

```python
#Using filter(), map(), and reduce():

#Extract transactions whose amount is greater than 5000.
#Add 18% GST to every selected transaction.
#Calculate the grand total after GST
transactions = [
    {"id": 1, "amount": 1200},
    {"id": 2, "amount": 5200},
    {"id": 3, "amount": 18000},
    {"id": 4, "amount": 2500},
    {"id": 5, "amount": 9600},
    {"id": 6, "amount": 7200}
]
from functools import reduce
transactions_5k=lambda x:x["amount"]<5000
transactions_2=filter(transactions_5k,transactions)
transactions_gst=map(lambda x:x["amount"]*1.18,transactions_2)
grand_total=reduce(lambda x,y:x+y,transactions_gst)
print(grand_total)
```

```python
#Using filter(), map(), and reduce():

#Extract active salespeople from the South region.
#Calculate a 5% incentive for each selected salesperson.
#Create a list containing salesperson name, sales amount, and incentive.
#Calculate the total incentive payable.
sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]
from functools import reduce
active_sales=lambda x:x["active"]==True and x["region"]=="South"
active_sales2=filter(active_sales,sales_records)
incentive=list(map(lambda x:{"name":x["salesperson"],"sales":x["sales"],"incentive":x["sales"]*0.05},active_sales2))
sales_incentive=reduce(lambda x,y:x+y["incentive"],incentive,0)
print(sales_incentive)
```
